# RC low-pass filter
A first-order filter: its corner frequency, its time constant, and what is left of a signal at one frequency. Change any value and everything that depends on it follows.

```python
R_1 := 10. kΩ
```

```python
C_1 := 100. nF
```

```python
V_in := 12.0 V
```

```python
f_c := 1/(2π R_1 C_1)
```

```python
τ := R_1 C_1
```

## Attenuation at one frequency

```python
f := 1.0 kHz
```

```python
V_out := V_in/√(1 + (f/f_c)²)
```

```python
A_dB := 20 log₁₀(V_out/V_in)  # gain in dB
```

## With component tolerances
`±` makes an interval; the worst case falls out of the same formula.

```python
R_tol := R_1 ± 5%
```

```python
C_tol := C_1 ± 10%
```

```python
f_tol := 1/(2π R_tol C_tol)
```

## Frequency response

```python
freqs := [10, 20, 50, 100, 200, 500, 1000, 2000, 5000, 10000] Hz;
gain := [20 log₁₀(1/√(1 + (x/f_c)²)) for x in freqs];
plot(freqs, gain, logx=True, grid=True, ylabel="gain [dB]", figsize=(6.4, 2.3))
```
