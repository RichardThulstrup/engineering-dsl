# Thermal check of the sensor board's 3.3 V regulator

**Purpose.** Can the linear regulator that makes the sensor board's 3.3 V from 5 V carry the board's load inside the enclosure without its junction getting too hot?

**References.**

1. The regulator's datasheet: SOT-223 package, 62 K/W from junction to ambient on a two-layer board, junction at most 125 °C.
2. *Power requirements: sensor board* (`power/sensor_board.ipynb`), for the load.

## Assumptions

- The air inside the enclosure is at most 55 °C.
- The board draws its full load all the time.
- The regulator's own quiescent current is left out: it is under 1 mA.

## Calculation

```python
# The regulator
V_in := 5.0 V
V_out := 3.3 V
I_out := 180 mA  # sensors and ADC, with room for a second sensor
```

```python
P_reg := (V_in - V_out) · I_out
```

```python
θ_JA := 62 K/W
T_amb := 55 °C
T_Jmax := 125 °C
```

```python
ΔT := θ_JA · P_reg
T_J := T_amb + ΔT
ΔT_margin := T_Jmax - T_J
```

## Result

The regulator turns about 0.31 W into heat, and its junction reaches about 74 °C: some 51 °C below its limit.

## Conclusion

The regulator in its SOT-223 package will do on a two-layer board, without extra copper. Twice the load would still keep its junction under 100 °C; at about 1.1 W it would reach its limit, and a switching regulator would be the better choice.
