## Own loads

```python
P_cpu := 5.0 V · 150 mA
P_display := 5.0 V · 120 mA
```

## The 5 V rail

```python
k_margin := 1.25  # design margin on the controller's own loads
P_5V := P_sensor + P_radio + k_margin · (P_cpu + P_display)
I_5V := P_5V / 5.0 V ▸ mA
I_5V_rated := 2.0 A  # rating of the buck converter
load_5V := I_5V / I_5V_rated
```

## Total

```python
V_supply := 24.0 V
η_buck := 0.90
P_controller := P_5V / η_buck
I_controller := P_controller / V_supply ▸ mA
```

The units keep their own margins; this sheet adds one only for its own loads. `P_controller` and `I_controller` are what the controller draws from 24 V.
