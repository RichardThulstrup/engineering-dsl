# Power requirements: radio module

The radio module takes 5 V from the controller and makes its 3.3 V with a small buck converter. The radio sends a tenth of the time and listens the rest. What the module draws is taken by *controller.ipynb*.

## Supply

```python
V_supply := 5.0 V
η_buck := 0.85  # 3.3 V buck converter at light load
```

## Loads

```python
P_tx := 3.3 V · 120 mA  # sending
P_rx := 3.3 V · 18 mA  # listening
d_tx := 10 %  # share of the time it sends
P_rf := d_tx · P_tx + (1 - d_tx) · P_rx
P_mcu := 3.3 V · 8 mA
```

## Total

```python
k_margin := 1.25  # design margin on the module's own loads
P_radio := k_margin · (P_rf + P_mcu) / η_buck
I_radio := P_radio / V_supply ▸ mA
```

`P_radio` and `I_radio` are what the module draws from 5 V, margin included.
