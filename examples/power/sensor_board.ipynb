# Power requirements: sensor board

The sensor board takes 5 V from the controller. A linear regulator makes 3.3 V for the sensors and the ADC; the status LEDs run from 5 V. What the board draws is taken by *controller.ipynb*.

## Supply

```python
V_supply := 5.0 V
η_LDO := 3.3 V / V_supply  # a linear regulator passes its current: η = V_out/V_in
```

## Loads

```python
# on 3.3 V
P_sensors := 3.3 V · 35 mA
P_adc := 3.3 V · 12 mA
# on 5 V
P_leds := 5.0 V · 10 mA
```

## Total

```python
k_margin := 1.25  # design margin on the board's own loads
P_sensor := k_margin · ((P_sensors + P_adc) / η_LDO + P_leds)
I_sensor := P_sensor / V_supply ▸ mA
```

`P_sensor` and `I_sensor` are what the board draws from 5 V, margin included.
