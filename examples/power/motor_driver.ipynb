# Power requirements: motor driver

The motor driver runs two motors from the 24 V bus; its logic has a regulator of its own. What the driver draws is taken by *system.ipynb*.

## Supply

```python
V_supply := 24.0 V
η_bridge := 0.95  # losses in the H-bridges
```

## Loads

```python
n_motors := 2
I_motor := 1.80 A  # each motor at its rated load
P_motors := n_motors · V_supply · I_motor
P_logic := 24.0 V · 15 mA  # the driver's logic
```

## Total

```python
k_margin := 1.25  # design margin on the driver's own loads
P_motor_drv := k_margin · (P_motors / η_bridge + P_logic)
I_motor_drv := P_motor_drv / V_supply
```

`P_motor_drv` and `I_motor_drv` are what the driver draws from 24 V, margin included.
