## The power supply

```python
k_margin := 1.25  # design margin on the system's own loads
P_24V := P_controller + P_motor_drv + k_margin · P_fan
I_24V := P_24V / 24.0 V
P_psu := 150 W  # rating of the chosen supply
load_psu := P_24V / P_psu
η_psu := 0.88
P_mains := P_24V / η_psu
```

`load_psu` is the share of its rating the supply carries, and `P_mains` is what it takes from the mains. When a unit's sheet changes, this sheet says which sheets are behind: *Edit ▸ Bring values up to date…* brings the whole chain along.
