# Power requirements: name of the unit

One sheet for each unit. List the unit's own loads below, and take the totals of the units it is made of from their sheets with *Edit ▸ Values from another sheet…*. This sheet's totals are what the sheet above it takes: name them after the unit (`P_unit`, `I_unit`), so that a sheet can take them from several units.

## Supply

```python
V_supply := 12.0 V
η_supply := 0.90  # efficiency of the unit's converter
```

## Loads

```python
P_logic := 3.3 V · 40 mA
P_io := 5.0 V · 20 mA
```

## Total

```python
k_margin := 1.25  # design margin on the unit's own loads
P_unit := k_margin · (P_logic + P_io) / η_supply
I_unit := P_unit / V_supply ▸ mA
```

`P_unit` and `I_unit` are what the unit draws from its supply, margin included.
